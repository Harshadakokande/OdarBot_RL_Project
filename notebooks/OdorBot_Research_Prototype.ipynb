# OdorBot Research Prototype

This notebook runs a compact, transparent study of sensory-history reinforcement learning for odor-source search. The primary research question is: does recent odor and wind information help an agent recover after a controlled wind-direction change?

This is a simplified prototype, not a validated biological or fluid-dynamics simulator.

```python
import sys
from pathlib import Path

repo_root = Path('.')
if (repo_root / 'src').exists():
    sys.path.insert(0, str(repo_root.resolve()))

print('Repository root:', repo_root.resolve())
```

```python
import json
from pathlib import Path

from src.environment import OdorSearchEnv
from src.experiment import run_experiment

config = {
    'grid_size': 9,
    'max_steps': 40,
    'training_episodes': 30,
    'evaluation_episodes': 12,
    'wind_change_step': 15,
    'history_window': 2,
    'seed': 7,
    'change_wind': True,
    'consecutive_detections': 2,
}

print('Configuration:')
print(json.dumps(config, indent=2))
```

```python
env = OdorSearchEnv(grid_size=config['grid_size'], max_steps=config['max_steps'], wind_change_step=config['wind_change_step'], seed=config['seed'])
obs = env.reset()
print('Initial observation:', obs)
print('Initial wind:', env.wind)
print('Source:', env.source)
```

```python
output_dir = Path('results')
output_dir.mkdir(exist_ok=True)
run_experiment(config, output_dir)
print(f'Run complete. Outputs in: {output_dir.resolve()}')
```

## Interpretation and limitations

A better result in this toy environment would suggest the history representation is useful under the chosen assumptions. It would not by itself validate the method for a real robot or a rich simulator. The response may reflect reward design, limited training, sparse tabular state coverage, or a proxy recovery metric rather than genuine tracking.
